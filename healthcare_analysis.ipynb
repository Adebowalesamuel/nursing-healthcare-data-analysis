# Nursing & Healthcare Data Analysis

### Exploring Patient Waiting Times and Satisfaction

This project demonstrates basic healthcare data analysis using Python and Pandas.

The analysis explores patient visits, waiting times, departments, visit types, and satisfaction scores using a fictional healthcare dataset.

**Tools:** Python • Pandas • Matplotlib • Google Colab • GitHub

> **Note:** The dataset is fictional and contains no real patient information.

## 1. Basic Data Analysis

The dataset is first analyzed using Python's built-in CSV handling and descriptive statistics.

```python
import urllib.request

url = "https://raw.githubusercontent.com/Adebowalesamuel/nursing-healthcare-data-analysis/main/data/healthcare_sample.csv"
urllib.request.urlretrieve(url, "healthcare_sample.csv")

print("Dataset downloaded successfully.")
import csv
from statistics import mean

file_path = "healthcare_sample.csv"

with open(file_path, newline="", encoding="utf-8") as file:
    data = list(csv.DictReader(file))

wait_times = [int(row["wait_time_minutes"]) for row in data]
satisfaction = [int(row["satisfaction_score"]) for row in data]

print("HEALTHCARE DATA ANALYSIS")
print("=" * 30)

print(f"Number of patient records: {len(data)}")
print(f"Average waiting time: {mean(wait_times):.1f} minutes")
print(f"Average satisfaction score: {mean(satisfaction):.1f}/5")
print(f"Longest waiting time: {max(wait_times)} minutes")
print(f"Shortest waiting time: {min(wait_times)} minutes")
!wget -O healthcare_sample.csv https://raw.githubusercontent.com/Adebowalesamuel/nursing-healthcare-data-analysis/main/data/healthcare_sample.csv
```

## 2. Exploring the Dataset with Pandas

Pandas is used to load the dataset into a DataFrame and explore the structure of the healthcare records.

```python
import pandas as pd

# Load the healthcare dataset
df = pd.read_csv("healthcare_sample.csv")

# Display the dataset as a table
df
```

## 3. Department-Level Analysis

The analysis compares average waiting times and satisfaction scores across departments.

```python
# Average waiting time by department

department_wait = (
    df.groupby("department")["wait_time_minutes"]
    .mean()
    .sort_values(ascending=False)
)

department_wait
```

```python
import matplotlib.pyplot as plt

# Create a bar chart of average waiting time by department
department_wait.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Average Waiting Time by Department")
plt.xlabel("Department")
plt.ylabel("Average Waiting Time (minutes)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Average satisfaction score by department

department_satisfaction = (
    df.groupby("department")["satisfaction_score"]
    .mean()
    .sort_values(ascending=False)
)

department_satisfaction
```

```python
import matplotlib.pyplot as plt

# Create a bar chart of average satisfaction by department
department_satisfaction.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Average Satisfaction Score by Department")
plt.xlabel("Department")
plt.ylabel("Average Satisfaction Score")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Key findings from the analysis

highest_wait_department = department_wait.idxmax()
highest_wait_time = department_wait.max()

lowest_wait_department = department_wait.idxmin()
lowest_wait_time = department_wait.min()

highest_satisfaction_department = department_satisfaction.idxmax()
highest_satisfaction_score = department_satisfaction.max()

lowest_satisfaction_department = department_satisfaction.idxmin()
lowest_satisfaction_score = department_satisfaction.min()

print("KEY FINDINGS")
print("=" * 40)

print(
    f"Highest average waiting time: "
    f"{highest_wait_department} "
    f"({highest_wait_time:.1f} minutes)"
)

print(
    f"Lowest average waiting time: "
    f"{lowest_wait_department} "
    f"({lowest_wait_time:.1f} minutes)"
)

print(
    f"Highest average satisfaction: "
    f"{highest_satisfaction_department} "
    f"({highest_satisfaction_score:.1f}/5)"
)

print(
    f"Lowest average satisfaction: "
    f"{lowest_satisfaction_department} "
    f"({lowest_satisfaction_score:.1f}/5)"
)
```
